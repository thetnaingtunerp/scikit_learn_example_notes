```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sklearn
from sklearn import preprocessing
```

```python
x = np.array([[100,200,350,200],[1,2,3,2]])
x
```

```python
#standardized
x_standardized = preprocessing.scale(x)
x_standardized
```

```python
#manually
x_std_manual = (x-np.mean(x,axis=0))/(np.std(x, axis=0))
x_std_manual
```

### fit and transform

```python
# fit and Transfrom
X_tranin = x
X_tranin
```

```python
scaler = preprocessing.StandardScaler().fit(X_tranin)
print(scaler.mean_)
print(scaler.var_)
```

```python
X_tranin_std = scaler.transform(X_tranin)
X_tranin_std
```

```python
print("Mean : ", np.mean(X_tranin_std, axis=0))
print("Standization : ", np.std(X_tranin_std, axis=0))
```
