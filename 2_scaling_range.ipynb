```python
import numpy as np
import sklearn
from sklearn import preprocessing
```

```python
#MinMaxScalar
X_train = np.array([[1,2],[3,4]])
X_test = np.array([[1,3],[2,4]])
```

```python
scalar =preprocessing.MinMaxScaler().fit(X_train)
print(scalar.data_min_)
print(scalar.data_max_)
```

```python
x_train_scale = scalar.transform(X_train)
x_train_scale
```

```python
x_test_scale = scalar.transform(X_test)
x_test_scale
```
