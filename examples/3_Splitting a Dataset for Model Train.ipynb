```python
import numpy as np
import pandas as pd

# Generate random data for our features (X) and target (y)
np.random.seed(42)  # For reproducibility
X = pd.DataFrame(np.random.rand(100, 2), columns=['feature_1', 'feature_2'])
y = pd.Series(np.random.randint(0, 2, 100)) # A binary target (0 or 1)

print("Shape of the original features (X):", X.shape)
print("Shape of the original target (y):", y.shape)
```

```python
from sklearn.model_selection import train_test_split

# Split the data into training and testing sets
# test_size=0.3 means 30% of the data will be for testing
# random_state=42 ensures the split is the same every time we run the code
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Inspect the Results
print("Shape of X_train:", X_train.shape)
print("Shape of X_test:", X_test.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of y_test:", y_test.shape)
```
