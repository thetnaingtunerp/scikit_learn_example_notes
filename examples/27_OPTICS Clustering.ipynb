```python
from sklearn.cluster import OPTICS
import numpy as np
import matplotlib.pyplot as plt

# ตัวอย่าง dataset
from sklearn.datasets import make_blobs
X, labels_true = make_blobs(n_samples=500, centers=3, cluster_std=0.5, random_state=0)

# OPTICS Model
clust = OPTICS(min_samples=10, xi=0.05, min_cluster_size=0.1)
clust.fit(X)

# Cluster labels
labels = clust.labels_

# Visualization
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='rainbow', s=10)
plt.title("OPTICS Clustering")
plt.show()
```
