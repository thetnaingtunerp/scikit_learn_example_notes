```python
#EDA
dataset = pd.read_csv('customer_reviews.csv', encoding='cp1252')
dataset['Rating'] = dataset['Rating'].str.split('out of ').str[0]
dataset['Rating'] = dataset['Rating'].astype(float)
dataset.isnull().sum()
```

```python
dataset = dataset.drop(columns=['Useful'])
```

```python
dataset['Date']=dataset['Date'].str.split('on ').str[1]
dataset[['Day', 'Month', 'Year']] = dataset['Date'].str.split(' ', expand=True)
```
