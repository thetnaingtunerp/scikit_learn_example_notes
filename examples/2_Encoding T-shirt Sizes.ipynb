##### Here is a step-by-step example project demonstrating the use of sklearn.preprocessing.LabelEncoder for ordinal data.

```python
import pandas as pd

data = {'Customer_ID': [1, 2, 3, 4, 5],
        'Size': ['Small', 'Medium', 'Large', 'Medium', 'Small'],
        'Purchase_Amount': [25, 30, 45, 30, 25]}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
```

```python
# Apply LabelEncoder
from sklearn.preprocessing import LabelEncoder

# Create an instance of the encoder
label_encoder = LabelEncoder()

# Fit and transform the 'Size' column
df['Size_Encoded'] = label_encoder.fit_transform(df['Size'])

print("\nDataFrame with Encoded 'Size' column:")
print(df)
```

```python
print("\nMapping from original labels to encoded integers:")
for i, label in enumerate(label_encoder.classes_):
    print(f"{label}: {i}")
```

```python
# Final DataFrame for Modeling
# Drop the original 'Size' column
final_df = df.drop('Size', axis=1)

print("\nFinal DataFrame for a machine learning model:")
print(final_df)
```

## Example 2

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

# Sample Data
data = pd.DataFrame({
    'Color': ['Red', 'Green', 'Blue', 'Red']
})

# Initialize encoder
encoder = OneHotEncoder()

# Fit and transform
encoded = encoder.fit_transform(data[['Color']])

# Convert to DataFrame for readability
encoded_df = pd.DataFrame(encoded.toarray(), columns=encoder.get_feature_names_out(['Color']))

# Combine with original data
final_df = pd.concat([data, encoded_df], axis=1)

print(final_df)
```
