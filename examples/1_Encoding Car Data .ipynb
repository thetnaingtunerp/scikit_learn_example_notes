### Create a Sample Dataset

```python
import pandas as pd

data = {'ID': [1, 2, 3, 4, 5],
        'Color': ['Red', 'Blue', 'Green', 'Red', 'Blue'],
        'Speed': [120, 150, 130, 140, 160]}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
```

```python
# Separate the Feature to be Encoded
# Separate the nominal feature to be encoded
nominal_data = df[['Color']]

# Separate the other features
other_features = df[['ID', 'Speed']]

print("\nNominal data to be encoded:")
print(nominal_data)
```

```python
#  Apply OneHotEncoder
from sklearn.preprocessing import OneHotEncoder

# Create an instance of the encoder
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')

# Fit and transform the data
encoded_data = encoder.fit_transform(nominal_data)

# The result is a NumPy array. Let's inspect it.
print("\nEncoded data (NumPy array):")
print(encoded_data)
```

```python
#  Convert Encoded Data Back to a DataFrame
# Get the new column names from the encoder
encoded_feature_names = encoder.get_feature_names_out(['Color'])

# Create a new DataFrame with the encoded data and column names
encoded_df = pd.DataFrame(encoded_data, columns=encoded_feature_names)

print("\nEncoded DataFrame:")
print(encoded_df)
```

```python
# Combine Encoded Data with Other Features
# Concatenate the original and encoded dataframes
final_df = pd.concat([other_features, encoded_df], axis=1)

print("\nFinal DataFrame for a machine learning model:")
print(final_df)
```
