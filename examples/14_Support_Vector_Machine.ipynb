Support Vector Machine (SVM) သည် supervised machine learning algorithm တစ်ခုဖြစ်ပြီး classification နှင့် regression problems များအတွက် အသုံးပြုနိုင်ပါသည်။ SVM ၏ အဓိကရည်ရွယ်ချက်မှာ data points များကို classes အလိုက် ခွဲခြားရန် optimal decision boundary (hyperplane) ကို ရှာဖွေရန်ဖြစ်သည်။

```python
# SVM classification Example 1
from sklearn import svm
from sklearn.model_selection import train_test_split
from sklearn import datasets
from sklearn.metrics import accuracy_score

# နမူနာ data များယူပါ
iris = datasets.load_iris()
X = iris.data[:, :2]  # ပထမ feature ၂ခုကိုပဲထားပါ
y = iris.target

# training and testing sets ခွဲပါ
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# SVM classifier ဖန်တီးပါ
clf = svm.SVC(kernel='linear')  # Linear kernel အသုံးပြုမည်

# Model ကို train လုပ်ပါ
clf.fit(X_train, y_train)

# Predictions လုပ်ပါ
y_pred = clf.predict(X_test)

# Accuracy စစ်ပါ
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.2f}")
```

### SVM တွင် kernel များစွာရှိပါသည်

```python
# Linear Kernel
linear_svm = svm.SVC(kernel='linear')

# Polynomial Kernel
poly_svm = svm.SVC(kernel='poly', degree=3)  # degree = polynomial degree

# RBF (Radial Basis Function) Kernel - အများဆုံးသုံးသော kernel
rbf_svm = svm.SVC(kernel='rbf', gamma='scale')  # gamma = kernel coefficient

# Sigmoid Kernel
sigmoid_svm = svm.SVC(kernel='sigmoid')
```

```python
# Hyperparameter Tuning Example
from sklearn.model_selection import GridSearchCV

# Hyperparameter grid သတ်မှတ်ပါ
param_grid = {
    'C': [0.1, 1, 10, 100],  # Regularization parameter
    'gamma': [1, 0.1, 0.01, 0.001],  # Kernel coefficient (RBF, poly အတွက်)
    'kernel': ['rbf', 'linear', 'poly']
}

# Grid search လုပ်ပါ
grid = GridSearchCV(svm.SVC(), param_grid, refit=True, verbose=2, cv=5)
grid.fit(X_train, y_train)

# Best parameters ကိုကြည့်ပါ
print(f"Best parameters: {grid.best_params_}")
print(f"Best estimator: {grid.best_estimator_}")

# Best model ဖြင့် prediction လုပ်ပါ
grid_predictions = grid.predict(X_test)
print(f"Grid search accuracy: {accuracy_score(y_test, grid_predictions):.2f}")
```

```python
# SVM for Regression (SVR) Exmaple
from sklearn.svm import SVR
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import StandardScaler

# နမူနာ data
X, y = datasets.make_regression(n_samples=100, n_features=1, noise=10, random_state=42)

# Data scaling (SVM သည် scaling ကိုအထူးဂရုစိုက်ရသည်)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Training and testing sets ခွဲပါ
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)

# SVR model ဖန်တီးပါ
svr = SVR(kernel='rbf', C=100, gamma=0.1, epsilon=0.1)

# Model ကို train လုပ်ပါ
svr.fit(X_train, y_train)

# Predictions လုပ်ပါ
y_pred = svr.predict(X_test)

# Performance စစ်ပါ
mse = mean_squared_error(y_test, y_pred)
print(f"Mean Squared Error: {mse:.2f}")
```
