```python
import numpy as np
from sklearn.linear_model import LinearRegression

# 1. Sample Data
months = np.array([1, 2, 3, 4, 5, 6])
sales = np.array([100, 150, 200, 250, 300, 350])

# Reshape the data for Scikit-learn
X = months.reshape(-1, 1)
y = sales

# 2. Train the model
model = LinearRegression()
model.fit(X, y)

# 3. Make predictions for future months
future_months = np.array([7, 8, 9]).reshape(-1, 1)
predicted_sales = model.predict(future_months)

print(f"Predicted sales for months 7, 8, and 9: {predicted_sales}")
# Output: Predicted sales for months 7, 8, and 9: [400. 450. 500.]
```
