```python
# 1. Missing Data ကို Detect လုပ်ခြင်း
import pandas as pd
import numpy as np

# Sample data
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
        'Age': [25, 30, np.nan, 40, 45],
        'Salary': [50000, 60000, 70000, np.nan, 90000]}
df = pd.DataFrame(data)

# Check for missing values
missing_values = df.isnull().sum()
print("Missing values in each column:")
print(missing_values)

```

```python
# 2. Scikit-Learn: SimpleImputer ကို အသုံးပြု၍ Missing Data ကို ဖြည့်စွက်ခြင်း
from sklearn.impute import SimpleImputer
# Create an imputer object with a mean filling strategy
imputer = SimpleImputer(strategy='mean')
# Fit the imputer on the DataFrame
imputer.fit(df[['Age', 'Salary']])
# Transform the DataFrame
df[['Age', 'Salary']] = imputer.transform(df[['Age', 'Salary']])
print("DataFrame after imputation:")
print(df)

```

```python
# 3. KNN Imputer ကို အသုံးပြု၍ Missing Data ကို ဖြည့်စွက်ခြင်း
from sklearn.impute import KNNImputer

# Create a KNN imputer object
knn_imputer = KNNImputer(n_neighbors=2)

# Fit the imputer on the DataFrame
knn_imputer.fit(df[['Age', 'Salary']])

# Transform the DataFrame
df[['Age', 'Salary']] = knn_imputer.transform(df[['Age', 'Salary']])
print("DataFrame after KNN imputation:")
print(df)
```

```python
# 4. Iterative Imputer ကို အသုံးပြု၍ Missing Data ကို ဖြည့်စွက်ခြင်း
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

# Create an iterative imputer object
iterative_imputer = IterativeImputer()

# Fit the imputer on the DataFrame
iterative_imputer.fit(df[['Age', 'Salary']])

# Transform the DataFrame
df[['Age', 'Salary']] = iterative_imputer.transform(df[['Age', 'Salary']])
print("DataFrame after Iterative imputation:")
print(df)
```
