```python
# 1. LabelEncoder
from sklearn.preprocessing import LabelEncoder
import numpy as np
import pandas as pd

# Sample data
data = {'Gender': ['Male', 'Female', 'Female', 'Male', 'Female'],
        'Age': [25, 30, 35, 40, 45],
        'Salary': [20000, 30000, 40000, 50000, 60000]}
df = pd.DataFrame(data)

# Encoding categorical features
label_encoder = LabelEncoder()
df['Gender'] = label_encoder.fit_transform(df['Gender'])

print(df)
```

```python
# 2. OneHotEncoder
from sklearn.preprocessing import OneHotEncoder

# Sample data
data1 = {'Gender': ['Male', 'Female', 'Female', 'Male', 'Female'],
        'Age': [25, 30, 35, 40, 45],
        'Salary': [20000, 30000, 40000, 50000, 60000]}
df = pd.DataFrame(data1)

# One-hot encoding
onehot_encoder = OneHotEncoder(sparse=False)
gender_encoded = onehot_encoder.fit_transform(df[['Gender']])
gender_encoded_df = pd.DataFrame(gender_encoded, columns=onehot_encoder.get_feature_names_out(['Gender']))

# Concatenate with original DataFrame
df = pd.concat([df, gender_encoded_df], axis=1)
df.drop('Gender', axis=1, inplace=True)

print(df)
```
