```python
from sklearn.feature_extraction.text import CountVectorizer

docs = ["I love machine learning", "Machine learning is amazing"]
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(docs)

print(vectorizer.get_feature_names_out())
print(X.toarray())


# Text ကို bag-of-words model အနေနဲ့ numeric matrix ပြောင်းပေးတယ်။
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

docs = ["I love machine learning", "Machine learning is amazing"]
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(docs)

print(vectorizer.get_feature_names_out())
print(X.toarray())

# TF-IDF (Term Frequency – Inverse Document Frequency) 
# algorithm ကို သုံးပြီး word importance ကို တိုင်းတာပေးတယ်။
# အနည်းဆုံးအကြိမ်သာ တွေ့တဲ့ (rare but informative) words တွေကို ပိုကြီးအောင် weighting ပေးတယ်။
```

```python

from skimage.feature import hog
from skimage import data, color
import matplotlib.pyplot as plt

image = color.rgb2gray(data.astronaut())
features, hog_image = hog(image, visualize=True)

plt.imshow(hog_image, cmap='gray')
plt.show()
```

```python
from sklearn.feature_extraction import FeatureHasher

data = [{"feature1": "apple", "feature2": "red"},
        {"feature1": "banana", "feature2": "yellow"}]

hasher = FeatureHasher(n_features=5, input_type='dict')
features = hasher.transform(data).toarray()
print(features)
# High-dimensional categorical data တွေအတွက် fast transformation လုပ်နိုင်တယ်။
```
