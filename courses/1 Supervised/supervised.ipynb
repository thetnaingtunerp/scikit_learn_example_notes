<h4>Supervised Learning ၏ အဓိကအမျိုးအစားများ</h4>

<ol start="1"><li><p><b>Classification (အမျိုးအစားခွဲခြားခြင်း):</b> ဒေတာများကို ကြိုတင်သတ်မှတ်ထားသော အမျိုးအစားများထဲသို့ ခွဲခြားခြင်းဖြစ်ပါသည်။</p><ul><li><p><b>ဥပမာ:</b> အီးမေးလ်တစ်စောင်သည် "spam" ဖြစ်သည် သို့မဟုတ် "inbox" ဖြစ်သည်ကို ခွဲခြားခြင်း၊ ဘဏ်ငွေလွှဲမှုတစ်ခုသည် "လိမ်လည်မှု (fraud)" ဟုတ်မဟုတ် ခွဲခြားခြင်း။</p></li><li><p><b>Scikit-Learn ရှိ Algorithms များ:</b></p><ul><li><p>Support Vector Machines (SVM)</p></li><li><p>K-Nearest Neighbors (KNN)</p></li><li><p>Logistic Regression</p></li><li><p>Decision Trees</p></li><li><p>Random Forest</p></li></ul></li></ul></li><li><p><b>Regression (ဆက်စပ်မှုခန့်မှန်းခြင်း):</b> ဒေတာများကြားရှိ ဆက်စပ်မှုကို သင်ယူ၍ ကိန်းဂဏန်းတန်ဖိုးတစ်ခုကို ခန့်မှန်းခြင်းဖြစ်ပါသည်။</p><ul><li><p><b>ဥပမာ:</b> အိမ်တစ်လုံး၏ အကျယ်အဝန်း၊ အိပ်ခန်းအရေအတွက် စသည့်အချက်များပေါ်မူတည်၍ ဈေးနှုန်းခန့်မှန်းခြင်း၊ ရာသီဥတုအခြေအနေပေါ်မူတည်၍ သီးနှံအထွက်နှုန်းခန့်မှန်းခြင်း။</p></li><li><p><b>Scikit-Learn ရှိ Algorithms များ:</b></p><ul><li><p>Linear Regression</p></li><li><p>Ridge Regression</p></li><li><p>Lasso Regression</p></li><li><p>Support Vector Regression (SVR)</p></li></ul></li></ul></li></ol>
