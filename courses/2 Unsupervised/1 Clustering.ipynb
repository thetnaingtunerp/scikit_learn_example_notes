| Algorithm                        | Description                                                                                                           |
| -------------------------------- | --------------------------------------------------------------------------------------------------------------------- |
| **KMeans**                       | Data ကို K clusters အနေနဲ့ ခွဲခြားပေးတယ်။ Cluster center (centroid) ကို iteratively update လုပ်ပြီး minimize လုပ်တယ်။ |
| **DBSCAN**                       | Density-based clustering. Cluster density နည်းတဲ့ point တွေကို noise အဖြစ် handle လုပ်တယ်။                            |
| **Agglomerative (Hierarchical)** | Bottom-up approach. Closest points တွေကို gradually merge လုပ်ပြီး tree-like structure ဖန်တီးတယ်။                     |
| **MeanShift**                    | Data density အရ mode (peak) ကို 찾아 cluster center တည်ဆောက်တယ်။                                                        |
