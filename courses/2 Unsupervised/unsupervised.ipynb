<ol start="1"><li><p><b>Clustering (အစုအဖွဲ့များခွဲခြားခြင်း):</b> ဒေတာများကို ၎င်းတို့၏ တူညီသော ဂုဏ်သတ္တိများပေါ်မူတည်၍ အစုအဖွဲ့များ (clusters) အဖြစ် ခွဲခြားခြင်းဖြစ်ပါသည်။</p><ul><li><p><b>ဥပမာ:</b> ဝယ်ယူသူများ၏ စျေးဝယ်ပုံစံကိုလေ့လာ၍ customer segment များခွဲခြားခြင်း၊ သတင်းဆောင်းပါးများကို အကြောင်းအရာအလိုက် အုပ်စုဖွဲ့ခြင်း။</p></li><li><p><b>Scikit-Learn ရှိ Algorithms များ:</b></p><ul><li><p>K-Means Clustering</p></li><li><p>DBSCAN</p></li><li><p>Hierarchical Clustering</p></li><li><p>Gaussian Mixture Models</p></li></ul></li></ul></li><li><p><b>Dimensionality Reduction (ဒေတာအရွယ်အစားလျှော့ချခြင်း):</b> ဒေတာများ၏ feature အရေအတွက် (dimensions) ကို အရေးကြီးသော အချက်အလက်များမပျောက်เสียစေဘဲ လျှော့ချခြင်းဖြစ်ပါသည်။ ၎င်းသည် model များ၏ သင်ယူမှုမြန်ဆန်စေရန်နှင့် data visualization ပြုလုပ်ရန် အသုံးဝင်ပါသည်။</p><ul><li><p><b>ဥပမာ:</b> Feature အများအပြားပါဝင်သော ဒေတာများကို 2D သို့မဟုတ် 3D ပုံများအဖြစ် ဖော်ပြရန် ပြောင်းလဲခြင်း။</p></li><li><p><b>Scikit-Learn ရှိ Algorithms များ:</b></p><ul><li><p>Principal Component Analysis (PCA)</p></li><li><p>t-Distributed Stochastic Neighbor Embedding (t-SNE)</p></li></ul></li></ul></li></ol>
